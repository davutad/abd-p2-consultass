```python
import pprint
from ODM import initApp
from consultas import CONSULTAS

initApp(db_name="envivo", scope=globals())

print("Modelos cargados:", [i for i in globals().keys() if i in ["recinto", "artista", "evento", "asistente"]])
```

```python
for cid, info in CONSULTAS.items():
    descripcion = info["descripcion"]
    nombre_coleccion = info["coleccion"]
    pipeline = info["pipeline"]

    print(f"\n{'='*70}")
    print(f"{cid} · {descripcion} (Colección: {nombre_coleccion})")
    print(f"{'='*70}")
    
    # Obtener la clase del modelo (artista, evento o recinto) dinámicamente
    Modelo = globals()[nombre_coleccion]
    
    # Ejecutar el aggregate a través de la clase del modelo del ODM
    resultado = list(Modelo.aggregate(pipeline))
    
    # Imprimir el resultado formateado
    pprint.pprint(resultado, sort_dicts=False)

```
